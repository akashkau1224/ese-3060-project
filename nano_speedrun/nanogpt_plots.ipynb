```python
import math
import matplotlib.pyplot as plt
```

```python
loss = [6.7356, 6.7335, 6.7256, 6.7287, 6.7226, 6.7239, 6.7231, 6.7061, 6.7070, 6.7183, 6.7079]
time = [242236, 244867, 249513, 249579, 265484, 249869, 265484, 268047, 262273, 275998, 270885]
loss2 = [6.7367, 6.7299, 6.7326, 6.7301, 6.7223, 6.7290, 6.7169, 6.7212, 6.7158, 6.7173, 6.7111]
time2 = [241604, 245409, 252541, 309587, 272187, 251328, 268427, 266163, 263320, 274413, 268871]
w = 0.96
function = []
for i in range(len(loss)):
    l = (loss[i] + loss2[i]) / 2
    t = (time[i] + time2[i]) / 2

    f = w*math.log(l) + (1-w)*math.log(t)
    f1 = w*math.log(loss[i]) + (1-w)*math.log(time[i])
    f2 = w*math.log(loss2[i]) + (1-w)*math.log(time2[i])
    print(f"{f1:.4f} {f2:.4f}")
    function.append(f)

x = [i/10.0 for i in range(30, 41)]

plt.figure(figsize=(10, 6))

plt.plot(x, function, marker='o', linestyle='-', color='tab:red', label='f(loss,time)')

plt.xlabel('Expansion Factor', fontsize=14)
plt.ylabel('f(loss, training time)', fontsize=14)
plt.title('Function vs Expansion Factor after 100 Iterations', fontsize=16)
plt.show()
```
